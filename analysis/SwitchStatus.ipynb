```python
import pandas as pd

bigdata = pd.read_csv('sw_traffics.csv', index_col=0)
if bigdata.index.name == 'ts':
    bigdata = bigdata.reset_index()

cols = ['td', 'sw', 'port', 'vl', 'send', 'receive']
print(bigdata[cols].tail().to_string(index=False))
```

```python
import numpy as np
import matplotlib.pyplot as plt

bigdata['td'] = pd.to_datetime(bigdata['td'], errors='coerce')
bigdata['tx_mb'] = bigdata['txb_raw'].fillna(0) / (1024**2)

bigdata['hour'] = bigdata['td'].dt.hour
bigdata['half'] = (bigdata['td'].dt.minute // 30) * 30
bigdata['hour_frac'] = bigdata['hour'] + bigdata['half'] / 60

per_30 = bigdata.groupby('hour_frac')['tx_mb'].mean().sort_index()

full_slots = np.arange(0, 24.5, 0.5)
per_30 = per_30.reindex(full_slots).interpolate(limit_direction='both')

mean_val = per_30.mean()

fig, ax = plt.subplots(figsize=(14, 5))

ax.set_xticks(range(24))
ax.set_xlim(0, 24)
ax.grid(axis='both', alpha=0.3)

for h in np.arange(0.5, 24, 1):
    ax.axvline(h, color='gray', linestyle=':', linewidth=0.8, alpha=0.7)

ax.plot(
    per_30.index, per_30.values,
    color='navy', linewidth=2.0,
    marker='o', markersize=4
)
ax.axhline(
    mean_val, color='black', linewidth=1.2,
    linestyle='--'
)

ax.set_xlabel('Hour of day')
ax.set_ylabel('Send traffic (MB)')
ax.set_title('Average send traffic over 24 hours')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

bigdata['td'] = pd.to_datetime(bigdata['td'], errors='coerce')
bigdata['tx_mb'] = bigdata['txb_raw'].fillna(0) / (1024**2)
bigdata['rx_mb'] = bigdata['rxb_raw'].fillna(0) / (1024**2)

bigdata['hour'] = bigdata['td'].dt.hour
bigdata['half'] = (bigdata['td'].dt.minute // 30) * 30
bigdata['hour_frac'] = bigdata['hour'] + bigdata['half'] / 60

send_30 = bigdata.groupby('hour_frac')['tx_mb'].mean().sort_index()
recv_30 = bigdata.groupby('hour_frac')['rx_mb'].mean().sort_index()

full_slots = np.arange(0, 24.5, 0.5)
send_30 = send_30.reindex(full_slots).interpolate(limit_direction='both')
recv_30 = recv_30.reindex(full_slots).interpolate(limit_direction='both')

mean_val = (send_30.mean() + recv_30.mean()) / 2

fig, ax = plt.subplots(figsize=(14, 5))

ax.set_xticks(range(24))
ax.set_xlim(0, 24)
ax.grid(axis='both', alpha=0.3)

for h in np.arange(0.5, 24, 1):
    ax.axvline(h, color='gray', linestyle=':', linewidth=0.8, alpha=0.7)

ax.plot(
    send_30.index, send_30.values,
    color='#2e75b6', linewidth=2.2,
    marker='o', markersize=2.5,
    markerfacecolor='black', markeredgecolor='black', markeredgewidth=0.3,
    label='Send'
)
ax.plot(
    recv_30.index, recv_30.values,
    color='#c0392b', linewidth=2.2,
    marker='o', markersize=2.5,
    markerfacecolor='black', markeredgecolor='black', markeredgewidth=0.3,
    label='Receive'
)

ax.axhline(
    mean_val, color='black', linewidth=1.2,
    linestyle='--',
    label='Mean (send & receive)'
)

ax.set_xlabel('Hour of day')
ax.set_ylabel('Traffic (MB)')
ax.set_title('Average send vs receive traffic over 24 hours')
ax.legend(loc='upper right', framealpha=0.9)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

bigdata['td'] = pd.to_datetime(bigdata['td'], errors='coerce')
bigdata['total_mb'] = (bigdata['txb_raw'].fillna(0) + bigdata['rxb_raw'].fillna(0)) / (1024**2)

bigdata['hour'] = bigdata['td'].dt.hour
bigdata['half'] = (bigdata['td'].dt.minute // 30) * 30
bigdata['hour_frac'] = bigdata['hour'] + bigdata['half'] / 60

per_30 = bigdata.groupby('hour_frac')['total_mb'].sum().sort_index()

full_slots = np.arange(0, 24.5, 0.5)
per_30 = per_30.reindex(full_slots).fillna(0)

mean_val = per_30.mean()

fig, ax = plt.subplots(figsize=(14, 5))

ax.set_xticks(range(24))
ax.set_xlim(0, 24)
ax.grid(axis='both', alpha=0.3)

for h in np.arange(0.5, 24, 1):
    ax.axvline(h, color='gray', linestyle=':', linewidth=0.8, alpha=0.7)

ax.plot(
    per_30.index, per_30.values,
    color='#2e75b6', linewidth=2.2,
    marker='o', markersize=2.5,
    markerfacecolor='black', markeredgecolor='black', markeredgewidth=0.3
)
ax.axhline(
    mean_val, color='black', linewidth=1.2,
    linestyle='--'
)

ax.set_xlabel('Hour of day')
ax.set_ylabel('Total traffic (MB)')
ax.set_title('Total traffic (send + receive) over 24 hours')
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

bigdata['td'] = pd.to_datetime(bigdata['td'], errors='coerce')
bigdata['hour'] = bigdata['td'].dt.hour
bigdata['day'] = bigdata['td'].dt.date.astype(str)
bigdata['tx_mb'] = bigdata['txb_raw'].fillna(0) / (1024**2)

pivot = bigdata.pivot_table(
    index='hour',
    columns='day',
    values='tx_mb',
    aggfunc='sum',
    fill_value=0
)

fig, ax = plt.subplots(figsize=(14, 7))
sns.heatmap(
    pivot,
    cmap='YlOrRd',
    annot=True,
    fmt='.2f',
    linewidths=0.5,
    linecolor='white',
    cbar_kws={'label': 'Total send (MB)'},
    ax=ax
)
ax.set_xlabel('Day')
ax.set_ylabel('Hour of day')
ax.set_title('Send traffic heatmap (hour × day, MB)')
ax.invert_yaxis()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

bigdata['td'] = pd.to_datetime(bigdata['td'], errors='coerce')
bigdata['ts_dt'] = pd.to_datetime(bigdata['ts'], unit='s', errors='coerce')

bigdata['dow'] = bigdata['ts_dt'].dt.dayofweek
bigdata['hour'] = bigdata['td'].dt.hour

bigdata['tx_mb'] = bigdata['txb_raw'].fillna(0) / (1024**2)
bigdata['rx_mb'] = bigdata['rxb_raw'].fillna(0) / (1024**2)
bigdata['total_mb'] = bigdata['tx_mb'] + bigdata['rx_mb']

agg = bigdata.groupby(['dow', 'hour'])['total_mb'].mean().reset_index()

pivot = agg.pivot(index='hour', columns='dow', values='total_mb')

order = [5, 6, 0, 1, 2, 3, 4]
labels = ['Sat', 'Sun', 'Mon', 'Tue', 'Wed', 'Thu', 'Fri']
pivot = pivot.reindex(columns=order)

fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(
    pivot,
    cmap='YlOrRd',
    annot=True,
    fmt='.2f',
    linewidths=0.5,
    linecolor='white',
    cbar_kws={'label': 'Avg traffic (MB)'},
    xticklabels=labels,
    ax=ax
)
ax.set_xlabel('Day of week')
ax.set_ylabel('Hour of day')
ax.set_title('Average traffic volume by day of week and hour (MB)')
ax.invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

bigdata['td'] = pd.to_datetime(bigdata['td'], errors='coerce')
bigdata['ts_dt'] = pd.to_datetime(bigdata['ts'], unit='s', errors='coerce')

last_ts = bigdata['ts_dt'].max()
cutoff = last_ts - pd.Timedelta(days=7)
recent = bigdata[bigdata['ts_dt'] >= cutoff].copy()

recent['dow'] = recent['ts_dt'].dt.dayofweek
recent['hour'] = recent['td'].dt.hour

agg = recent.groupby(['dow', 'hour'])['send'].mean().reset_index()
pivot = agg.pivot(index='hour', columns='dow', values='send')

order = [5, 6, 0, 1, 2, 3, 4]
labels = ['Sat', 'Sun', 'Mon', 'Tue', 'Wed', 'Thu', 'Fri']
pivot = pivot.reindex(columns=order)

fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(
    pivot,
    cmap='YlOrRd',
    annot=True,
    fmt='.1f',
    linewidths=0.5,
    linecolor='white',
    cbar_kws={'label': 'Avg send'},
    xticklabels=labels,
    ax=ax
)
ax.set_xlabel('Day of week')
ax.set_ylabel('Hour of day')
ax.set_title(f'Average send traffic by day of week and hour (last 7 days)')
ax.invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

bigdata['td'] = pd.to_datetime(bigdata['td'], errors='coerce')
bigdata['ts_dt'] = pd.to_datetime(bigdata['ts'], unit='s', errors='coerce')
last_ts = bigdata['ts_dt'].max()
cutoff = last_ts - pd.Timedelta(days=7)
recent = bigdata[bigdata['ts_dt'] >= cutoff].copy()

by_sw = recent.groupby('sw')[['send', 'receive']].sum().reset_index()
by_sw['total'] = by_sw['send'] + by_sw['receive']
by_sw = by_sw.sort_values('total', ascending=False)

top_n = 15
top = by_sw.head(top_n).iloc[::-1]

fig, ax = plt.subplots(figsize=(10, max(4, top_n * 0.45)))
y = range(len(top))
h = 0.4

ax.barh([i + h/2 for i in y], top['send'], height=h, color='steelblue', label='Send', edgecolor='black')
ax.barh([i - h/2 for i in y], top['receive'], height=h, color='seagreen', label='Receive', edgecolor='black')

ax.set_yticks(list(y))
ax.set_yticklabels(top['sw'])
ax.set_xlabel('Total traffic (bytes)')
ax.set_ylabel('Switch (IP)')
ax.set_title(f'Top {top_n} switches by send/receive (last 7 days)')
ax.legend()
ax.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()
```
